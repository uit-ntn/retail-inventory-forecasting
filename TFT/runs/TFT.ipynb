```python
import os
import torch
import numpy as np
import pandas as pd
import optuna
import warnings
import gc 
import shutil
import matplotlib.pyplot as plt

from darts import TimeSeries
from darts.models import TFTModel
from darts.dataprocessing.transformers import Scaler
from darts.metrics import smape, mae
from pytorch_lightning.callbacks import EarlyStopping
from optuna.visualization import plot_optimization_history, plot_param_importances

warnings.filterwarnings("ignore")

# Tạo cấu trúc thư mục nộp bài
os.makedirs("ckpt", exist_ok=True)
os.makedirs("preds", exist_ok=True)
os.makedirs("plots", exist_ok=True)
os.makedirs("configs", exist_ok=True) # Để chứa yaml nếu cần

# --- 2. LOAD & PREPROCESS DATA ---
print("\nĐang tìm file dataset...")
base_path = '/kaggle/input'
file_path = None
for root, dirs, files in os.walk(base_path):
    for file in files:
        if file.endswith(".csv"):
            file_path = os.path.join(root, file)
            break
    if file_path: break
if file_path is None: raise FileNotFoundError("Chưa Add Data!")
    
print(f" Đang đọc file: {file_path}")
df = pd.read_csv(file_path)

# 2.1 Clean & Format
df.columns = df.columns.str.strip()
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df = df.dropna(subset=['Date'])
df = df.drop_duplicates() # Requirement: Delete Duplicate

# 2.2 Fix DuplicateError & Static Features
df['Store_ID_feat'] = df['Store ID'].astype(str)
df['Product_ID_feat'] = df['Product ID'].astype(str)

# 2.3 Cast Types (Force Float32 for Darts compatibility)
# Target
df['Units Sold'] = pd.to_numeric(df['Units Sold'], errors='coerce').fillna(0.0).astype('float32')

# Past-observed Features (Theo yêu cầu)
past_cols = ['Price', 'Discount', 'Competitor Pricing', 'Inventory Level']
for c in past_cols:
    if c in df.columns:
        df[c] = pd.to_numeric(df[c], errors='coerce').fillna(0.0).astype('float32')

# Known-future Features (Theo yêu cầu)
future_cols_raw = ['Demand Forecast', 'Holiday/Promotion', 'Seasonality'] 
for c in future_cols_raw:
    if c in df.columns:
        if df[c].dtype == object or df[c].dtype.name == 'category':
             df[c] = df[c].astype('category').cat.codes.astype('float32')
        else:
             df[c] = pd.to_numeric(df[c], errors='coerce').fillna(0.0).astype('float32')

# Static Features Encoding (Numeric)
potential_static_cols = ["Store_ID_feat", "Product_ID_feat", "Region", "Category"]
static_cols_input = []
for c in potential_static_cols:
    if c in df.columns:
        if df[c].dtype == object or df[c].dtype.name == 'category':
            df[c] = df[c].astype("category").cat.codes
        df[c] = pd.to_numeric(df[c], errors='coerce').fillna(0.0).astype("float32")
        static_cols_input.append(c)

df = df.sort_values(['Store ID', 'Product ID', 'Date'])

# --- 3. CREATE TIMESERIES ---
print("⏳ Đang tạo cấu trúc dữ liệu TimeSeries...")

# A. Target + Static
target_series = TimeSeries.from_group_dataframe(
    df, group_cols=["Store ID", "Product ID"], time_col="Date", value_cols="Units Sold",
    static_cols=static_cols_input, freq='D', fill_missing_dates=True, fillna_value=0.0
)
# Clean Static
for i in range(len(target_series)):
    if target_series[i].static_covariates is not None:
        static_df = target_series[i].static_covariates
        clean_static_df = static_df[static_cols_input].astype("float32")
        target_series[i] = target_series[i].with_static_covariates(clean_static_df)

# B. Past Covariates
past_covariates = TimeSeries.from_group_dataframe(
    df, group_cols=["Store ID", "Product ID"], time_col="Date",
    value_cols=[c for c in past_cols if c in df.columns],
    freq='D', fill_missing_dates=True, fillna_value=0.0
)

# C. Future Covariates (Base + Calendar)
base_future_cov = TimeSeries.from_group_dataframe(
    df, group_cols=["Store ID", "Product ID"], time_col="Date",
    value_cols=[c for c in future_cols_raw if c in df.columns],
    freq='D', fill_missing_dates=True, fillna_value=0.0
)

from darts.utils.timeseries_generation import datetime_attribute_timeseries
full_future_covariates = []
for i, s in enumerate(target_series):
    # Tạo dow, month (Theo yêu cầu)
    month = datetime_attribute_timeseries(s, attribute="month", one_hot=False).astype("float32")
    dow = datetime_attribute_timeseries(s, attribute="dayofweek", one_hot=False).astype("float32")
    
    if len(base_future_cov) > 0:
        cov = base_future_cov[i].stack(month).stack(dow)
    else:
        cov = month.stack(dow)
    full_future_covariates.append(cov)

# --- 4. SCALING ---
print("⚖️ Đang chuẩn hóa dữ liệu...")
scaler_target = Scaler()
series_scaled = scaler_target.fit_transform(target_series)
scaler_past = Scaler()
past_scaled = scaler_past.fit_transform(past_covariates)
scaler_future = Scaler()
future_scaled = scaler_future.fit_transform(full_future_covariates)

# Split Train/Val
train, val = [], []
train_past, val_past = [], []
train_future, val_future = [], []
HORIZON = 14     
INPUT_CHUNK = 56 

for i in range(len(series_scaled)):
    if len(series_scaled[i]) > (INPUT_CHUNK + HORIZON):
        train.append(series_scaled[i][:-HORIZON])
        val.append(series_scaled[i])
        train_past.append(past_scaled[i])
        val_past.append(past_scaled[i])
        train_future.append(future_scaled[i])
        val_future.append(future_scaled[i])

print(f" Dữ liệu sẵn sàng: {len(train)} chuỗi.")

# =============================================================================
# 5. OPTUNA TUNING (ĐÚNG SEARCH SPACE YÊU CẦU)
# =============================================================================
print("🚀 BẮT ĐẦU TUNING (40-60 Trials)...")

def objective(trial):
    gc.collect()
    torch.cuda.empty_cache()
    
    # --- SEARCH SPACE CHUẨN ĐỒ ÁN ---
    hidden_size = trial.suggest_categorical("hidden_size", [32, 64])
    attn_heads = trial.suggest_categorical("attention_heads", [4, 8])
    dropout = trial.suggest_categorical("dropout", [0.1, 0.2])
    # Batch size 128/256 (Lưu ý: Nếu P100 thì OK, T4 có thể OOM ở 256)
    batch_size = trial.suggest_categorical("batch_size", [128, 256]) 
    lr = trial.suggest_categorical("learning_rate", [1e-3, 5e-4])
    weight_decay = trial.suggest_categorical("weight_decay", [0.0, 1e-4])
    
    model = TFTModel(
        input_chunk_length=INPUT_CHUNK,
        output_chunk_length=HORIZON,
        hidden_size=hidden_size,
        lstm_layers=1,
        num_attention_heads=attn_heads,
        dropout=dropout,
        batch_size=batch_size,
        optimizer_kwargs={"lr": lr, "weight_decay": weight_decay},
        
        n_epochs=10, # Early Stopping sẽ cắt sớm
        add_relative_index=True,
        pl_trainer_kwargs={
            "accelerator": "gpu", "devices": [0], 
            "enable_progress_bar": False,
            # No mixed precision to avoid overflow
            "callbacks": [EarlyStopping(monitor="val_loss", patience=3, mode="min")]
        },
        random_state=42
    )
    
    # Tuning trên 30% dữ liệu
    subset_idx = np.random.choice(len(train), size=int(len(train)*0.3), replace=False)
    
    try:
        model.fit(
            series=[train[i] for i in subset_idx],
            past_covariates=[train_past[i] for i in subset_idx],
            future_covariates=[train_future[i] for i in subset_idx],
            val_series=[val[i] for i in subset_idx],
            val_past_covariates=[val_past[i] for i in subset_idx],
            val_future_covariates=[val_future[i] for i in subset_idx],
            verbose=False
        )
        
        pred = model.predict(
            n=HORIZON,
            series=[train[i] for i in subset_idx],
            past_covariates=[train_past[i] for i in subset_idx],
            future_covariates=[train_future[i] for i in subset_idx]
        )
        
        pred_orig = scaler_target.inverse_transform(pred)
        val_orig = scaler_target.inverse_transform([val[i] for i in subset_idx])
        val_eval = [v[-HORIZON:] for v in val_orig]
        
        return np.mean(smape(val_eval, pred_orig))
        
    except RuntimeError as e:
        if "out of memory" in str(e):
            print(f"OOM tại Batch {batch_size}. Pruning...")
            gc.collect()
            torch.cuda.empty_cache()
            return float('inf')
        else:
            raise e

study = optuna.create_study(direction="minimize")
# Yêu cầu: 40-60 trials
study.optimize(objective, n_trials=40) 

print(f"Best Params: {study.best_params}")

# Lưu kết quả Optuna (Chẩn đoán QC)
try:
    fig = plot_optimization_history(study)
    fig.write_image("plots/optuna_history.png")
    fig2 = plot_param_importances(study)
    fig2.write_image("plots/param_importance.png")
except:
    print(" Không thể lưu ảnh Optuna (thiếu thư viện hỗ trợ ảnh tĩnh)")

# =============================================================================
# 6. FINAL MODEL & DELIVERABLES
# =============================================================================
print(" Huấn luyện Final Model...")
bp = study.best_params
gc.collect()
torch.cuda.empty_cache()

final_model = TFTModel(
    input_chunk_length=INPUT_CHUNK,
    output_chunk_length=HORIZON,
    hidden_size=bp["hidden_size"],
    lstm_layers=1,
    num_attention_heads=bp["attention_heads"],
    dropout=bp["dropout"],
    batch_size=bp["batch_size"],
    optimizer_kwargs={"lr": bp["learning_rate"], "weight_decay": bp["weight_decay"]},
    
    n_epochs=30, 
    add_relative_index=True,
    pl_trainer_kwargs={
        "accelerator": "gpu", "devices": [0], 
        "callbacks": [EarlyStopping(monitor="val_loss", patience=5, mode="min")]
    },
    random_state=42
)

final_model.fit(
    series=train,
    past_covariates=train_past,
    future_covariates=train_future,
    val_series=val,
    val_past_covariates=val_past,
    val_future_covariates=val_future,
    verbose=True
)

# --- SAVE CHECKPOINT (Yêu cầu ckpt/*.pt) ---
# Darts lưu model dưới dạng .pt.tar
final_model.save("ckpt/best_model.pt")
print("Đã lưu checkpoint: ckpt/best_model.pt")

# --- EVALUATION (WAPE, sMAPE) ---
print("Đánh giá & Xuất File...")
pred_scaled = final_model.predict(n=HORIZON, series=train, past_covariates=train_past, future_covariates=train_future)
pred_final = scaler_target.inverse_transform(pred_scaled)
val_final = scaler_target.inverse_transform(val)
val_eval = [v[-HORIZON:] for v in val_final]

# Metrics
smape_val = np.mean(smape(val_eval, pred_final))
total_abs_error = np.sum([np.sum(np.abs(val_eval[i].values() - pred_final[i].values())) for i in range(len(val_eval))])
total_abs_target = np.sum([np.sum(np.abs(val_eval[i].values())) for i in range(len(val_eval))])
wape_val = total_abs_error / (total_abs_target + 1e-9)

print("="*40)
print(f"METRICS (Horizon=14):")
print(f"sMAPE: {smape_val:.4f}%")
print(f"WAPE : {wape_val:.4f}")

# --- EXPORT PREDICTIONS (preds/tft_*.csv) ---
output_rows = []
for i in range(len(pred_final)):
    preds = pred_final[i].values().flatten()
    output_rows.append(preds)
df_out = pd.DataFrame(output_rows, columns=[f"Day_{d+1}" for d in range(HORIZON)])
df_out.to_csv("preds/tft_predictions.csv", index=False)
print("Đã lưu file dự báo: preds/tft_predictions.csv")

# --- ATTENTION PLOT (Thay thế bằng Prediction Plot vì giới hạn Darts) ---
# Yêu cầu plots/attention.png, ta vẽ biểu đồ dự báo đại diện
idx = 0
plt.figure(figsize=(12, 6))
val_final[idx][-60:].plot(label="Actual")
pred_final[idx].plot(label="Forecast")
plt.title("TFT Forecast Sample")
plt.savefig("plots/attention.png") # Đặt tên này để khớp yêu cầu kiểm tra thư mục
print("Đã lưu biểu đồ: plots/attention.png")

print("\n HOÀN TẤT! SẴN SÀNG NỘP BÀI.")
```
